TransactionId items
t1             Bread,Milk
t2             Bread,Diaper,Beer,Egg
t3             Milk,Diaper,Beer,Coke
t4             Bread,Milk,Diaper,Beer
t5             Bread,Milk,Diaper,coke
find the frequent itemset.

```python
transactions = [
    ['Bread', 'Milk'],
    ['Bread', 'Diaper', 'Beer', 'Egg'],
    ['Milk', 'Diaper', 'Beer', 'Coke'],
    ['Bread', 'Milk', 'Diaper', 'Beer'],
    ['Bread', 'Milk', 'Diaper', 'Coke']
]
min_support = 2
def get_support(itemset):
    count = 0
    for transaction in transactions:
        if set(itemset).issubset(set(transaction)):
            count += 1
    return count
items = set()
for transaction in transactions:
    for item in transaction:
        items.add(item)
items = sorted(items)
def generate_candidates(previous_itemsets, k):
    candidates = []
    for i in range(len(previous_itemsets)):
        for j in range(i + 1, len(previous_itemsets)):
            union = sorted(
                set(previous_itemsets[i]) |
                set(previous_itemsets[j])
            )
            if len(union) == k and union not in candidates:
                candidates.append(union)
    return candidates
frequent = []
for item in items:
    support = get_support([item])
    if support >= min_support:
        frequent.append([item])
print("Frequent 1-itemsets:")
for itemset in frequent:
    print(itemset, "->", get_support(itemset))
k = 2
while frequent:
    candidates = generate_candidates(frequent, k)
    new_frequent = []
    for itemset in candidates:
        support = get_support(itemset)
        if support >= min_support:
            new_frequent.append(itemset)
    if not new_frequent:
        break
    print("\nFrequent", k, "-itemsets:")
    for itemset in new_frequent:
        print(itemset, "->", get_support(itemset))
    frequent = new_frequent
    k += 1
```
