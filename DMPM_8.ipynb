```python
import pandas as pd
url="https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv"
data =pd.read_csv(url, sep=";")
print("Dataset shape:",data.shape)
print("\nFirst 5 rows:")
print(data.head())
x =data["alcohol"]
y =data["quality"]
covariance =x.cov(y)
correlation =x.corr(y)
print("RESULT: ")
print("Covariance between Alcohol and Quality:",round(covariance,4))
print("Correlation coefficient between Alcohol and Quality:",round(correlation,4))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.scatter(
    data["alcohol"],
    data["quality"],
    alpha=0.4
)
plt.xlabel("Alcohol")
plt.ylabel("Wine Quality")
plt.title("Alcohol vs Wine Quality")
plt.show()
```
