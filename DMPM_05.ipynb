**WAP to implement z-score noemalization,min-max, and decimal scalinng normaliztion on(10,20,30,40,0)**

```python
import math
data=[10,20,30,40,50]
mean= sum(data)/len(data)
sum_square = 0
for x in data:
  sum_square+=(x-mean)**2
std=math.sqrt(sum_square/len(data))
z_score = []
for x in data:
  z_score.append((x-mean)/std)
min_value = min(data)
max_value = max(data)
min_max = []
for x in data:
  value =(x-min_value)/(max_value-min_value)
  min_max.append(value)
decimal = []
for x in data:
  decimal.append(x/100)
print("Data: ",data)
print("Z-score: ")
print([round(x,2)for x in z_score])
print("Min-Max: ")
print([round(x,2) for x in min_max])
print("Decimal scaling: ")
print([round(x,2) for x in decimal])
```

WAP to implement the decimal scaling normalization on a dataset.

```python
import pandas as pd
data = pd.read_csv("Iris.csv")
x= data["sepalLengthCm"]
max_value=max(abs(x))
j = len(str(int(max_value)))
normalized = x/(10**j)
print("Original value: ",x.head(10))
print("After normalization : ",normalized.head(10))
```

WAP to implement min_max normalization on any dataset and display the normalized values.

```python
import pandas as df
data=df.read_csv("Titanic_dataset.csv")
age = data["Age"].dropna()
min_value = age.min()
max_value = age.max()
normalized = (age-min_value)/(max_value-min_value)
print("Original value: ",age.head(10))
print("After normalization : ",normalized.head(10))
```
