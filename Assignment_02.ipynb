WAP to find the median of data points with using functions.


```python
def median():
  a = int(input("Enter the number of data points: "))
  data = []
  for i in range(a):
    num = float(input("Enter the number: "))
    data.append(num)
  data.sort()
  if a%2==1:
    med= data[a//2]
  else:
    med = (data[a//2-1]+data[a//2])/2
  print("Median:",med)
median()
```

```python
# WAP to find standard deviation using function

def standard_deviation():
    n = int(input("Enter the number of data points: "))
    data = []

    for i in range(n):
        num = float(input("Enter data point: "))
        data.append(num)

    total = 0
    for i in range(n):
        total = total + data[i]

    mean = total / n

    sum_sq = 0
    for i in range(n):
        sum_sq = sum_sq + (data[i] - mean) ** 2

    variance = sum_sq / n
    sd = variance ** 0.5

    print("Mean =", mean)
    print("Variance =", variance)
    print("Standard Deviation =", sd)

standard_deviation()

# WAP to find standard deviation without using function

n = int(input("Enter the number of data points: "))
data = []
for i in range(n):
    num = float(input("Enter data point: "))
    data.append(num)
total=0
for i in range(n):
    total =total data[i]
mean =total/n
sum_sq =0
for i in range(n):
    sum_sq=sum_sq+(data[i]-mean)**2
variance=sum_sq/n
sd=variance**0.5
print("Mean =", mean)
print("Variance =", variance)
print("Standard Deviation =", sd)
```
